```python
import pandas as pd

#create a DataFrame
sales_df = {
    'Date':['04-10-2025', '12-03-2025', '15-06-2024', '04-07-2025'],
    'Store':['Apple_Store', 'Best_Buy', 'Home_Depot', 'Walmart'],
    'Product':['Laptop', 'Phone', 'Office_table', 'Printer'],
    'Quantity':[4, 15, 2, 4],
    'Price':[1000, 500, 800, 200]
}

df = pd.DataFrame(sales_df)
print("\nPandas DataFrame")
print(df)
```

```python
#Getting summary information of the DataFrame
summary = df.describe()
print("\nSummary Information:")
print(summary)
```

```python
#include all columns
df.describe(include='all')
```

```python
# Adding a new column to the DataFrame
df['Total_sales'] = df['Quantity'] * df['Price']
print("\nDataFrame after adding a new column:")
df
```

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df.info()
```

```python
#Filter the DataFrame to get all sales that occurred 
# in the month of July.
july_sales = df[df['Date'].dt.month_name() == 'July']
print(july_sales)

#or using month number
july_sale = df[df['Date'].dt.month == 7]
print(july_sale)
```

```python
#Get all sales where the Quantity sold is greater than 10.
sales_filter = df[df['Quantity'] > 10]

print("\nFiltering data where Quantity > 10:")
print(sales_filter)
```

```python
#Get the unique store names then compute 
# the total sales per store.
unique_store_names = df['Store'].unique()
print(unique_store_names)
```

```python
#compute the total sales per store.
df['Total-store_sale'] = df['Quantity'] * df['Price']
df
```

```python
#Get the unique product and calculate 
# the average price of each product.
unique_product = df['Product'].unique()
print(unique_product)
```

```python
#the average price of each product.
average_product_price = df.groupby('Product')['Price'].mean()
print(average_product_price)
```

```python
#Sort the DataFrame by Total Sales in descending order.
sorted_df = df.sort_values(by='Total_sales', ascending=False)
print(sorted_df)
```

```python
#Rank the stores based on Total Sales
df['Total_Sales'] = df['Quantity'] * df['Price']

store_sales = df.groupby('Store')['Total_Sales'].sum().reset_index()

store_sales['Rank'] = store_sales['Total_Sales'].rank(ascending=False,
    method='dense'
)

print(store_sales)
```

```python
#Define custom function
#discounted_price = ['Price'] * 0.1
#final_price = ['Price'] - discounted_price
#final_price = Price - (0.1 × Price)
#final_price = Price × (1 - 0.1)
##final_price = Price × 0.9
def discounted_price(row):
    if row['Quantity'] > 5:
       return row['Price'] * 0.9
    return row['Price']
#Apply the custom function to the DataFrame
df['Discounted_Price'] = df.apply(discounted_price, axis=1)
df
print(df)
```
