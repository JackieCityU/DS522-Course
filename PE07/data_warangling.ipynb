```python
import pandas as pd
import numpy as np

# Step 1: Load the datasets
movies_df = pd.read_csv('movies 2.csv')
ratings_df = pd.read_excel('ratings.xlsx')
#ratings_df
movies_df
```

```python
ratings_df
```

```python
#inspect data first
print(movies_df.isnull().sum())

#checking datatype and basic information
print(ratings_df.info())
```

```python
#inner join
inner_join = pd.merge(movies_df,ratings_df, on="movieId",
    how="inner"
)
inner_join
```

```python
#left join
left_join = pd.merge(movies_df, ratings_df, on="movieId", 
how="left")
left_join
```

```python
#outer join
outer_join = pd.merge(movies_df, ratings_df, on="movieId", 
how="outer")
outer_join
```

```python
#right join
right_join = pd.merge(movies_df, ratings_df,on="movieId",
    how="right"
)
right_join
```

```python
# Create bins
bins = [1, 2, 3, 4, 5]
# Assign ratings to bins
ratings_df["digitized_rating"] = np.digitize(
    ratings_df["rating"],
    bins
)
ratings_df
```

```python
ratings_df["rating_category"] = pd.cut(
   ratings_df["rating"],
    bins=[0, 2, 3.5, 4.5, 5],
    labels=["Low", "Medium", "High", "Very High"]
)
ratings_df
```

```python
ratings_df["rating_quantile"] = pd.qcut(
   ratings_df["rating"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)
ratings_df
```

```python
ratings_df = ratings_df.fillna(0)
ratings_df
```

```python
ratings_df['rating'] = ratings_df['rating'].fillna(
    ratings_df['rating'].mean())
ratings_df
```

```python
ratings_df.isnull().sum()
```

```python
#stack and unstack
#we first create a pivot
ratings_wide = ratings_df.pivot(index="userId", columns="movieId",
    values="rating"
)
ratings_wide
```

```python
stacked_df = ratings_wide.stack()
stacked_df
```

```python
unstacked_df = stacked_df.unstack()
unstacked_df
```

```python
#Finding most rated genres
#Will first merge the two dfs
merged_df = pd.merge(movies_df, ratings_df, on="movieId")
merged_df
```

```python
#split genres into lists
merged_df["genres"] = merged_df["genres"].str.split("|")
merged_df
```

```python
#Exploding genres into rows
exploded_df = merged_df.explode("genres")
exploded_df
```

```python
#Finding genre with highest number of ratings
genre_ratings = exploded_df.groupby("genres")["rating"].count()
genre_ratings.sort_values(ascending=False)
```

```python
#Top genre
top_genre = genre_ratings.idxmax()
top_genre
```
