```python
#merging the movies and ratings 
# DataFrames on the movieId column.
merged_df = pd.merge(movies_df, ratings_df, on='movieId')
merged_df
```

```python
#or we can also do this
movie_ratings_merge = movies_df.merge(ratings_df, on="movieId")
movie_ratings_merge
```

```python
#inspecting insights, getting highest rated movie
top_movies = merged_df.groupby("title")["rating"].mean()
print(top_movies.sort_values(ascending=False))
```

```python
#getting the most popular movie
popular_movie = merged_df['title'].value_counts()
print(popular_movie)
```

```python
#get average rating by genre
genre_ratings = merged_df.groupby("genres")["rating"].mean()
print(genre_ratings)
```

```python
# getting ratings by most active users
active_users = merged_df.groupby("userId").value_counts()
print(active_users)
```

```python
#Cleaning the datasets
#checking for missing values first, if missing, return True, if not missing, return False
print(merged_df.isnull())
```

```python
#Let me count missing values per column
print(merged_df.isnull().sum())
```

```python
clean_df = merged_df.dropna()
clean_df
```

```python
#Convert data types as necessary (convert timestamp to a readable datetime format).
merged_df["timestamp"] = pd.to_datetime(merged_df["timestamp"], unit="s")
merged_df
```

```python
#Extract the release year from the title column and 
# creating a new column release_year.
merged_df["release_year"] = merged_df["title"].str.extract(r"\((\d{4})\)")
merged_df
```

```python
#In pandas, we use vectorization operations because they can be faster especially with 
#larger datasets, if we use loops, they can be slow as i will demonstrate below;
movie_highest_rating = merged_df.groupby("title").agg(
    average_rating=("rating", "mean"),
    rating_count=("rating", "count")
)
filtered_movies = movie_highest_rating[movie_highest_rating["rating_count"] >= 100]

top_movie = filtered_movies.sort_values(
    by="average_rating",
    ascending=False
)
print("\nThe movie that has the highest average rating is: ")
print(top_movie.head(1))
```

```python
#if we go for the 'for loop' optiono, it is so slow:
#using unique(), mean(), filtering()
movie_avg_ratings = []

for movie in merged_df["title"].unique():
    movie_data = merged_df[merged_df["title"] == movie]
    rating_count = movie_data["rating"].count()
    
    if rating_count >= 100:
        avg_rating = movie_data["rating"].mean()
        movie_avg_ratings.append((movie, avg_rating))

top_movie = sorted(movie_avg_ratings, key=lambda x: x[1], reverse=True)[0]
print("\nThe movie that has the highest average rating is: ", top_movie)
print(top_movie)
```

```python
#groupby() and agg(),this is much faster
movie_higher_rate = merged_df.groupby("title").agg(
    average_rating=("rating", "mean"),
    rating_count=("rating", "count")
)
filtered_movies = movie_higher_rate[movie_higher_rate["rating_count"] >= 100]

top_movie = filtered_movies.sort_values(
    by="average_rating",
    ascending=False
)
print("\nThe movie that has the highest average rating is: ")
print(top_movie.head(1))
```

```python
#genre of movies tends to get the highest ratings
# Separate genres into individual rows
genres_df = merged_df.copy()
genres_df["genres"] = genres_df["genres"].str.split("|")
genres_df = genres_df.explode("genres")

# Calculate average rating and number of ratings per genre
genre_ratings = genres_df.groupby("genres").agg(
    average_rating=("rating", "mean"),
    rating_count=("rating", "count")
)

# Sort genres by highest average rating
top_genres = genre_ratings.sort_values(
    by="average_rating",
    ascending=False
)
print("\nThe genres of movies that tends to have the highest rating is: ")
print(top_genres)
```

```python
#What is the average rating for movies released each year? 
#Option 1: use unique(), and mean().
year_results = []
for year in merged_df["release_year"].unique():
    year_data = merged_df[merged_df["release_year"] == year]
    avg_rating = year_data["rating"].mean()
    year_results.append((year, avg_rating))

year_ratings_option1 = pd.DataFrame(
    year_results,
    columns=["release_year", "average_rating"]
)
year_ratings_option1 = year_ratings_option1.sort_values(
    by="average_rating",
    ascending=False
)
print(year_ratings_option1)
```

```python
#Option 2: use groupby() and agg().
year_ratings_option2 = merged_df.groupby("release_year").agg(
    average_rating=("rating", "mean")
)
year_ratings_option2 = year_ratings_option2.sort_values(
    by="average_rating", 
    ascending=False
)
print(year_ratings_option2)
```

```python
#Apply Custom Functions:
#Use the apply() function with a lambda to create a new column 
#that categorizes movies as "Old" if they were released before 2000, and "New" otherwise.
merged_df["movie_category"] = merged_df["release_year"].apply(
    lambda x: "Unknown" if pd.isna(x)
    else "Old" if int(x) < 2000
    else "New"
)
merged_df
```

```python
#Use map() to apply this mapping to the genres column.
merged_df["main_genre"] = merged_df["genres"].str.split("|").str[0]

genre_map = {
    "Comedy": "Popular",
    "Drama": "Classic",
    "Action": "Exciting"
}

merged_df["genre_label"] = merged_df["main_genre"].map(genre_map)

merged_df[["main_genre", "genre_label"]]
```
