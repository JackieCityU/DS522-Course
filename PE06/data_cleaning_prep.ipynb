```python
import pandas as pd
import numpy as np

df = pd.read_csv('retail_sales.csv', encoding='latin')
df
```

```python
#inspect data first
print(df.isnull().sum())

#checking datatype and basic information
print(df.info())
```

```python
#filling missing customer ids with unknown
df["CustomerID"] = df["CustomerID"].fillna("Unknown")

df
```

```python
#dropping rows with missing total amount
df = df.dropna(subset=["TotalAmount"])
df
```

```python
#checking for duplicate rows
print(df.duplicated().sum())
```

```python
df.duplicated()
```

```python
#View duplicate rows
print(df[df.duplicated()])
```

```python
#remove duplicated rows
df = df.drop_duplicates()
df
```

```python
#Replace negative Quantity values with 0
df.loc[df["Quantity"] < 0, "Quantity"] = 0

#Adjusting TotalAmount accordingly
df["TotalAmount"] = df["Quantity"] * df["Price"]

print(df.head())
```

```python
#checking data types
print(df.dtypes)
```

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    origin="1899-12-30",
    unit="D"
)

df
```

```python
print(df["CustomerID"].dtype)
```

```python
#customerID as a string
df["CustomerID"] = df["CustomerID"].astype(str)
df
```

```python
print(df["CustomerID"].dtype)
```

```python
#Adding new column month
df["Month"] = df["Date"].dt.month
df
```

```python
#Add a new column RevenueCategory that categorizes transactions based on TotalAmount:
#"Low" for transactions under $50
#"Medium" for transactions between $50 and $200
#"High" for transactions above $200
df["RevenueCategory"] = pd.cut(df["TotalAmount"],
bins=[0, 50, 200, float("inf")],
labels=["Low", "Medium", "High"]
)
df

```

```python
#Filter out transactions where Price or Quantity is 0.
df = df[
    (df["Price"] !=0) & (df["Quantity"] != 0)
    ]
df
```

```python
#Convert all product names to title case (e.g., product a -> Product A).
df["Product"] = df["Product"].str.title()
df
```

```python
#Remove any leading or trailing whitespace from the product names.
df["Product"] = df["Product"].str.strip()
df
```

```python
#Clean up the CustomerID column:
#Ensure that all CustomerID values are in uppercase.
df["CustomerID"] = df["CustomerID"].str.upper()
df
```

```python
#Remove any leading or trailing whitespace from the CustomerID.
df["CustomerID"] = df["CustomerID"].str.strip()
df
```

```python
#If a CustomerID starts with the string "TEMP", replace it with "CUST".
df["CustomerID"] = df["CustomerID"].str.replace("TEMP", "CUST")
df
```

```python
#Create a new column ProductCategory that categorizes products based on their names.
#Extract information from the Product column:
#Create a new column ProductCategory that categorizes products based on their names. For example:
#If the product name contains "A" or "B", categorize it as "Category 1".
#If the product name contains "C" or "D", categorize it as "Category 2".

df["ProductCategory"] = df["Product"].apply(
    lambda x: "Category 1"
    if("A" in x or "B" in x)
    else "Category 2"
    if("C" in x or "D" in x)
    else "other"
)
df
```

```python
#Save the cleaned and prepared dataset to a new file called “cleaned_retail_sales.csv”.
df.to_csv("cleaned_retail -sales.csv", index=False)

#we use index=False tto avoid Pandas from saving row numbers as an extra column, unwanted columns.

```
